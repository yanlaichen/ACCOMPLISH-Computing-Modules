# Activity 1: Variables & Data Types


##Variable
Python is a dynamically typed language, which means you don't need to declare a variable's type — Python figures it out for you automatically.

A **variable** is simply a name that refers to a value stored in memory. Python supports several basic data types, including:

*   int – whole numbers (e.g. 2, 10, -5)
*   float – decimal numbers (e.g. 3.14, -0.5)
*   str – strings of text (e.g. "physics", "velocity")
*   bool – boolean values: True or False

**Let’s look at some examples:**

````python
# Integer: a whole number
x = 5
y = 10
# Float: a number with decimals
velocity = 3.5

# String: text inside quotation marks
name = "physics"

# Boolean: True or False
is_physics_fun = True

print(x+y)
````

# Activity 2: Lists, Slicing, and Loops

## **Lists**

A list is a collection of items (like numbers or words). You can store multiple values in one list and access them using their position (called an *index*).

*Index in python starts at 0 instead of 1!*


**Example Below:**

````python
# Create a list of displacements (in meters)
displacements = [0, 2, 4, 6, 8]

# Print the whole list
print("Full list:", displacements)

# Print just the first 3 elements using slicing
print("First 3 displacements:", displacements[-3]) #start at index 0 and stop before index 3
````

In Python, you can use negative indexing to access items from the end of a list.



*   `-1` refers to the last item
*   `-2` refers to the second-to-last item, and so on

**See the example below**


````python
displacements = [0, 2, 4, 6, 8]

print("Last displacement:", displacements[-1])
````

## **Loops**
A loop lets you tell the computer:

> “Do this again and again for every item in a list.”

In Python, loops let you repeat actions automatically. There are two main kinds of loops:


*   For loops – loop through items in a list or range
*   While loops – keep looping as long as a condition is true

### **For Loop:**

This is the most common kind of loop. It’s perfect when you already know what you're looping through, like a list of numbers or names.

**Let's look at an example below**

````python
displacements = [0, 2, 4, 6, 8]


# Loop through each value in the list
for x in displacements:
    # It is crucial that you indent the body of the loop.
    # This way, Python knows that's the code you want to loop!
    print("Displacement:", x)
````

The `d` above is a variable or a temporary name.

Think of it like a placeholder:

Each time the loop runs, `d` takes the value of the next item in the list.

The computer does this for every item, so `d` will be `0` the first time, then `2`, then `4`, and so on.

**Take a look at the other example below and try changing name of the placeholder variable**



````python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
    print("Fruit:", x)
````

### **For Loop with `range()`**:

If you want to repeat something a specific number of times, use `range()`.

From below:

`range(5)` means go from 0 to 4 (5 numbers total)

**Keep in mind** that index starts from 0!

````python
# Repeat something 5 times
for i in range(5):
    print("This is loop number:", i)
````

**You can also start at a different number:**

````python
for i in range(1, 6):  # Starts at 1, ends at 5
    print(i)
````

### **While Loop**:

A while loop keeps going as long as a condition is true.

Be careful — if the condition never becomes false, it will loop forever!

````python
# Loop while a number is less than 5
count = 0

while count < 5:
    print("Count is:", count)
    count = count + 1
````

**Try deleting `count = count + 1` from the code above and see what happens**

### **Quick Review**


| Loop Type          | Use When You Want To…                  | Example Syntax        |
| ------------------ | -------------------------------------- | --------------------- |
| `for x in list`    | Loop through each item in a list       | `for name in names:`  |
| `for i in range()` | Repeat something a set number of times | `for i in range(10):` |
| `while condition:` | Keep looping while a condition is true | `while x < 5:`        |

**Common mistakes to watch out for**


*   If you forget to update the variable in a while loop, it might never stop!
*   `range(n)` starts at 0 by default
* Be sure to use indentation (spacing) to show what is "inside" the loop




# Activity 3: Functions and Comments

**What is a Function?**

A function is a reusable block of code that does a specific job. You can give it inputs (called parameters) and it will return an output.

**What is a Comment?**

A comment is a line in your code that Python ignores. It’s just for humans to read — used to explain what your code does.

You write a comment using `#`:

````python
# This is a comment. Python won't run this line.
````

In physics, you may remember the formula:

***Final position = Initial position + Velocity × Time***

Let’s write a function that does that:

````python
# Define the function
def final_position(initial, velocity, time):
    # This function calculates final position using: d = d0 + vt
    return initial + velocity * time

# Call the function with sample values
print("Final position:", final_position(0, 3, 5))  # Should return 15

print("final position: ", final_position(1, 5, 10))
````

# Activity 4: Arrays (Using NumPy)

**What Is an Array?**

An array is like a list — it holds many values — but it’s better for math and science, especially when you’re working with vectors, measurements, or large datasets.

Python has a powerful library called NumPy that makes arrays faster and easier to use for calculations.

To use NumPy, we first import it:

````python
import numpy as np
````

Now we can make a NumPy array using `np.array()`

````python
# Create a NumPy array
displacement_array = np.array([0, 2, 4, 6, 8])

print("Array:", displacement_array)
````

The great thing with arrays is that you can do math to the arrays! Let's look at some examples below.

````python
# Add a constant to every value
print(displacement_array + 2)  # adds 2 to every number

# Multiply all values by a number
print(displacement_array * 10)  # scales everything
````

NumPy arrays also work great for physics vectors!

````python
# Define 2D vectors
v1 = np.array([3, 4])
v2 = np.array([1, 2])

# Add the vectors
v_sum = v1 + (v2*-1)
print("v2 =", v2*-1)
print("v1 + v2 =", v_sum)

# Magnitude (length) of a vector
print("Magnitude of v1:", np.linalg.norm(v_sum))
#norm() is a function that finds the length (magnitude) of a vector
````
