# Activity 1: Boolean Expressions and Conditional Statements

Last class, you learned about the Boolean type, which can be `True` or `False`. **Boolean expressions** are expressions that result in a Boolean value.

Let's see some examples:

```python
velocity = 5

print("Velocity > 0:", velocity > 0)
print("Velocity <= 0:", velocity <= 0)
print("Velocity = 5:", velocity == 5)

print() # Add an empty line to make the output pretty :)

print("Setting velocity to 10...")
velocity = 10
print("Velocity = 5:", velocity == 5)
print("Velocity != 5:", velocity != 5)
```

Most of these operators look familiar from math. Be careful – the math `=` is `==` in Python! Remember, `=` is for assigning values to variables, not for equality.

In addition to our familiar math operators, Python also uses words (like `and`, `or`) as operators, too. It's as if we're writing actual English in our code!

Here are examples for each of these keywords. Before you run the code block, see if you can guess Boolean output!

```python
# velocity = 10
print(velocity)
print("Velocity > 0 and Velocity < 20:", velocity > 0 and velocity < 20)

print("Velocity < 0 or Velocity > 5:", velocity < 0 or velocity > 5)

print("Not Velocity > 0:", not velocity > 0)

my_list = [1, 2, 3, 4, 5]
print("3 in my_list:", 3 in my_list)
print("6 not in my_list:", 6 not in my_list)

a = [1, 2]
b = [1, 2]
c = a
print("a is b:", a is b)
print("a is c:", a is c)
print("a is not b:", a is not b)
```

```python
a = "hello"
b = "hello"

print(a is b)
```

What do these keywords mean?
- `and` and `or` combine expressions. `and` returns `True` only if both expressions are `True`. `or` returns `True` if one of the expressions is `True`.
- `not` returns the opposite of the expression. `True` becomes `False`, and `False` becomes `True`.
- `in` checks if the value is in the list
- `is` checks if the two values literally the same Python variable. This is slightly different from `==`, which compares for equality. Most of the time, this means the same thing, but not always! Be careful!

## Conditional Statements

In Boolean expressions, you can use numbers, strings, variables, and even call functions. They are become over-powered when combined with conditional statements (aka `if`/`else` statements).

With `if`/`else` statements, you can write some code to run when the condition is true and some other code to run when the condition is false.


```python
velocity = 0  # m/s

if velocity > 0:
    print("The object is moving forward.")
elif velocity < 0: # short for "else if"
    print("The object is moving backward.")
else:
    print("The object is at rest.")
```

Since `velocity` is 5, the `if` code runs, and none of the other code.

# Activity 2: More on Lists



## Useful List Tips

- `len(my_list)`: Gives you the number of items in a list
- `my_list.append(v)`: Appends items to the end of your list
- `my_list + other_list`: Concatenates two lists together

Let's see them in action!

```python
my_list = []

print("my_list before:", my_list)

for i in range(5):
  my_list.append(i)

print("my_list after:", my_list)
print("my_list has", len(my_list), "items")

print("Adding 6 to my_list...")
my_list.append(6)
print(my_list)
print("my_list has", len(my_list), "items")

other_list = [7, 8, 9]
print("Concatenating other_list to my_list...")
my_list = my_list + other_list
print("my_list has", len(my_list), "items")

for i in range(len(my_list)):  # len() is useful for setting the range
  print("Item", i, "is", my_list[i])
```

## List Comprehension - the *cool* way to make lists

List comprehension allows you to make new lists in just one line of code. It essentially uses a single-line for-loop.

In the previous example, we used a for-loop to create `my_list`.

```python
my_list = []

for i in range(5):
  my_list.append()

print(my_list)
```

With list comprehension, we can create `my_list` in one line.

```python
my_list = ["Number" + str(i) for i in range(5)]
print(my_list)
```

Notice how the `for i in range(5)` is the same in both pieces of code. List comprehension basically says:
> Add `i` to the list for every number in `range(5)` (0 to 4)

*Note*: We use square brackets to indicate to Python that we're creating a list.

# Activity 3: Tuples, Sets, and Dictionaries

## Tuples

A tuple is a collection of items that is ordered and immutable (cannot be changed after creation). Tuples are defined using parentheses (), and the items are separated by commas.

Here's an example:

```python
my_tuple = (1, 2, 3, "hello")
print(my_tuple)
print(my_tuple[0])  # Accessing elements by index
my_tuple[1] = 5  # This will raise a TypeError because tuples are immutable
```

Just like lists, tuples can contain items of any type. But, the difference is that tuples cannot be modified.

The most common example of using tuples is to represent pairs of related information (e.g. x-y coordinates, index-item pairs).

See the example code below.

```python
point1 = (10, 20)
point2 = (-5, 15)

print(f"Point 1 coordinates: {point1}")
print(f"Point 2 x-coordinate: {point2[0]}")

# You can even assign variables to items in a tuple, like this
x1, x2 = point1
print(f"Point 1 x-coordinate: {x1}")
print(f"Point 1 y-coordinate: {x2}")
```

> Note: the `f` denotes an **f-string**. It helps you to write string templates, where you can write Python inside your string by wrapping the code in `{}`. Don't worry about fully understanding this right now! You'll see many more examples of this throughout this course.

## Sets

A **set** is a collection of unique items. Sets never have duplicates! You can create a set using `{}` or `set()`.

```python
rainbow = {"red", "orange", "yellow", "green", "blue", "indigo", "violet"}
# rainbow = set(["red", "orange", "yellow", "green", "blue", "indigo", "violet"])
print(rainbow)
print(type(rainbow))
```

You can add elements.

```python
primary_colors = set() # an empty set
primary_colors.add("red")
primary_colors.add("yellow")
primary_colors.add("blue")
print(primary_colors)
```

You can remove elements, too.

```python
primary_colors.remove("yellow")
primary_colors.add("green")
print(primary_colors)
```

Adding duplicates doesn't change the set.

```python
primary_colors.add("green")
print(primary_colors)
```

## Dictionaries

A **dictionary** is a collection of items, indexed by keys, instead of indices. Keys are unique identifiers that you use to access the corresponding values. Keys can only be a string, number, or tuple; however, values can be anything you'd like!

Let's see some examples!

```python
car_dict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}
print(car_dict)
```

You can access the items of a dictionary by referring to its key name, inside square brackets.

```python
x = car_dict["model"]
print(x)
```

You can change the value of a specific item by referring to its key name.

```python
car_dict["year"] = 2018
print(car_dict)
```

You can loop through a dictionary by using a `for` loop.

```python
for key, value in car_dict.items():
  print(key, value)

print() # Add an empty line to make the output pretty :)

for key in car_dict.keys():
  print(key)

print() # Add an empty line to make the output pretty :)

for value in car_dict.values():
  print(value)
```

You can check if a key exists in a dictionary by using the `in` keyword.

```python
if "model" in car_dict:
  print("Yes, 'model' is one of the keys in the my_dict dictionary")
```

# Activity 4: Objects and Dataclasses

So, now you've seen most of Python's built-in types. These are called primitives. You can also create objects, which are custom Python things. Objects are created using the `class` keyword. They require an `__init__` function, but can contain many other functions that you write!

Here's an example of a calculator class:

```python
class Calculator:
  def __init__(self):
    self.result = 0

  def add(self, x):
    self.result += x

  def subtract(self, x):
    self.result -= x

  def multiply(self, x):
    self.result *= x

  def divide(self, x):
    self.result /= x

  def clear(self):
    self.result = 0

  def get_result(self):
    return self.result
```

For this class, we will primarily be working with **dataclasses**. **Dataclasses** are object that just contain data. They can be a convenient way to represent lots of related data (we can them **attributes**). To use dataclasses, we must import them from the `dataclass` library.

Let's see an example!

```python
from dataclasses import dataclass

@dataclass
class Point:
  x: int
  y: int

@dataclass
class OneDMotion:
  position: Point
  velocity: float # aka a decimal number
  time: float

# Example usage
start_pos = Point(x=0, y=0)
motion_state = OneDMotion(position=start_pos, velocity=5.0, time=0.0)

print(f"Initial state: Position: ({motion_state.position.x}, {motion_state.position.y}), Velocity: {motion_state.velocity}, Time: {motion_state.time}")
```
