# Activity 1: Simple Plots

**What is Matplotlib?**

Matplotlib is a popular Python library for creating graphs and plots.

Before we can use the Matplotlib plotting tools, we need to tell Python to load that library as seen below.

```python
import matplotlib.pyplot as plt
```



*   `matplotlib.pyplot` is a module that contains all the functions to make plots
*   We give it a short nickname `plt` so it's easier to type later

To use matplotlib, we can use the dot `.` to access a function inside a module or object.

So when you write: `plt.plot(x, y)`

You are saying, “From the `plt` toolbox, use the function called `plot`.”

Let's plot a simple set of 2D vector data. This simulates what you might collect from a physics experiment measuring acceleration over time.

```python
import matplotlib.pyplot as plt

# Sample data
time = [0, 1, 2, 3, 4, 5]
acceleration = [0, 2, 4, 6, 5, 3]

# Line plot (connects the dots)
plt.plot(time, acceleration)

# Show the plot
plt.show()
```

## What is a Line Plot?

Line plot, like we see above, connects each data point with a straight line which is great for showing change over time or trends.




## What is a Scatter Plot?

A scatter plot is like a Line Plot but you just want to see the individual data points.

**Let's take a look at an example below**

```python
# Scatter plot (shows just the points)
plt.scatter(time, acceleration)

plt.show()
```

** When Do We Use Scatter vs. Line?**

| Plot Type       | Best For                                                                               |
| --------------- | -------------------------------------------------------------------------------------- |
| `plt.plot()`    | Showing **trends** or continuous change                                                |
| `plt.scatter()` | Showing **individual data points**, especially if the relationship is unclear or noisy |


# Activity 2: Labels and Titles in Plots

Now that you can draw a simple plot using `plt.plot()`, let’s improve the appearance and clarity of your graph. Clear, labeled plots are super important in physics and science reporting!

Let's add:


*   X and Y axis labels — to explain what your data represents
*   A title — to tell people what the whole graph is about
* Grid lines — to make your graph easier to read

 **Let's take a look at an example below: **

```python
# Sample data
time = [0, 1, 2, 3, 4, 5]
acceleration = [0, 2, 4, 6, 5, 3]

# Create the plot
plt.plot(time, acceleration, color='red', marker='o', linestyle='--')

# Add labels and title
plt.xlabel("Time (seconds)")
plt.ylabel("Acceleration (m/s²)")
plt.title("Acceleration vs. Time")

# Turn grid on
plt.grid(True)

# Show the plot
plt.show()
```

If you want to see what parameters each function takes, you can search online. For example `plt.plot(x,y)` works but [here](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html) we can see that you also can take in a variety of `Line2D` properties like `color`, `fillstyle` and `animated`.


Some new styles introduced above that you can use are:

| Function               | What it Does                     |
| ---------------------- | -------------------------------- |
| `color='blue'`         | Changes line color               |
| `marker='o'`           | Adds a circle at each data point |
| `linestyle='--'`       | Makes the line dashed            |
| `plt.grid(True/False)` | Turns grid on or off             |


**Other `plt` styling function you can try:**

# Activity 3: Reading simple CSV files

## **What is Pandas in Python?**

Pandas is a Python library used to store, explore, and analyze data — kind of like a spreadsheet in code.

You use Pandas when you want to:
* Read data from a CSV or Excel file
* Clean up messy data
* Look at trends in a dataset
* Make graphs (with help from Matplotlib)


Pandas stands for "Panel Data", which means datasets with rows and columns — just like tables or spreadsheets.


**Let's import it below:**


```python
import pandas as pd
```

## **What is a CSV File?**

CSV stands for Comma-Separated Values — it’s a common way to store data in table form. You can open one with Excel, Google Sheets, or Python!


**Let's load a real dataset below about Air Travel**

***Our goal is to only display the months from 1958 with more than 350 passengers.***

```python
# Load the CSV from a URL
url = 'https://people.sc.fsu.edu/~jburkardt/data/csv/airtravel.csv'
data = pd.read_csv(url)

# Show raw column names
print("Original column names:")
print(data.columns)

# Clean column names
data.columns = data.columns.str.replace('"', '').str.strip()
print(data.columns)

# Preview the first few rows
print("\n First 5 rows of the dataset:")
print(data.head())

# Show just the 1958 column
print("\n Passenger counts for 1958:")
print(data["1958"])

# Filter months with more than 350 passengers in 1958
high_passengers = data[data["1958"] > 350]
print("\n Months with more than 350 passengers in 1958:")
print(high_passengers)

# Only keep Month and 1958 columns
filtered_1958 = high_passengers[["Month", "1958"]]
print("\n Months with more than 350 passengers in 1958:")
print(filtered_1958)
```

`data = pd.read_csv()` loads the data into something called a DataFrame — a table where:


*   **rows** = each data record  (in this case, each month)
*   **columns** = Each category (years like 1958, 1959, etc.)


---


`print(data.columns)` This prints the original column names. Sometimes, CSV files contain unwanted spaces or quotation marks around column names — this can break your code if you try to access a column like "1958" but it’s actually stored as ' "1958" '.

---

`data.columns = data.columns.str.replace('"', '').str.strip()`

The above code cleans up the column names by:


*  Removing any double quotes (`"`)
*  Stripping any extra spaces before or after


Now you can access the columns reliably using their names, like `"1958"`.

---


`data.head()` Shows the first 5 rows of the data which is super useful to quickly check what’s inside.

---


`print(data["1958"])` This line selects just the "1958" column : the number of passengers each month in that year.

---


`high_passengers = data[data["1958"] > 350]`
This filters the data. You’re telling Pandas:
“Only keep the rows where the number of passengers in 1958 is greater than 350.” The result is a new DataFrame called `high_passengers` with only those months.


You will notice that prints the full rows for those months — including 1959 and 1960 columns too (since they’re still part of the table).

---


`filtered_1958 = high_passengers[["Month", "1958"]]`

This step selects only the "`Month`" and "`1958`" columns from the filtered data — so the output is simpler and easier to read.

Now lets try adding a column where we map the House to a Trait.

**(House → Trait)**

```python
# Map houses to traits
house_traits = {
    "Griffindor": "Bravery",
    "Slytherin": "Cunning",
    "Ravenclaw": "Wisdom",
    "Hufflepuff": "Kindness"
}

# Add new column
df["Trait"] = df["House"].map(house_traits)

# Preview updated DataFrame
print("\n🏷️ Data with Traits:")
print(df)
```

## Plot Trimmed Data

Using what we learned from previous activities to plot the original data and the trimmed data.



```python
import matplotlib.pyplot as plt

# Plot
plt.bar(df["Name"], df["Classes_Attended"], color="goldenrod")
plt.xlabel("Student Name")
plt.ylabel("Classes Attended")
plt.title("Hogwarts Students with High Attendance")
plt.grid(True)
plt.show()

# Plot
plt.bar(active_students["Name"], active_students["Classes_Attended"], color="green")
plt.xlabel("Student Name")
plt.ylabel("Classes Attended")
plt.title("Hogwarts Students with High Attendance")
plt.grid(True)
plt.show()
```
