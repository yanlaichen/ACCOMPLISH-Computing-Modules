# Case Study 2: Predictive Maintenance Classification

This notebook connects supervised learning, classification, class imbalance, and model evaluation using a predictive-maintenance setting.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

plt.rcParams['figure.figsize'] = (10, 4)
````

## Task 1: Load a predictive-maintenance dataset
Preferred online source: UCI AI4I 2020 Predictive Maintenance Dataset. If internet access is unavailable, use the included AI4I-like classroom sample.

````python
# Option A: local classroom sample
df = pd.read_csv('../data/case_study_2_ai4i_like_predictive_maintenance_sample.csv')
df.head()
````

## Task 2: Define features and target
Here the target is `Machine_failure`. This is a classification problem because the output is a class label.

````python
features = ['Air_temperature_K','Process_temperature_K','Rotational_speed_rpm','Torque_Nm','Tool_wear_min']
X = df[features]
y = df['Machine_failure']
print(y.value_counts())
````

## Task 3: Train/test split and baseline model

````python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)
clf = RandomForestClassifier(n_estimators=100, random_state=42, class_weight='balanced')
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred, zero_division=0))
````

## Task 4: Interpret feature importance

````python
importances = pd.Series(clf.feature_importances_, index=features).sort_values(ascending=False)
importances.plot(kind='bar')
plt.ylabel('Importance')
plt.title('Random Forest Feature Importance')
plt.show()
importances
````
